# Тапсырма 6. Жаңа бағандар
EDA бағытыңызға пайдалы кемінде екі жаңа баған құрыңыз: біреуі бар бағандардан есептелген
сан (мысалы, қатынас, айырма немесе қосынды), екіншісі сандық бағанды `pd.cut()` немесе
шарт арқылы санаттарға бөлу. Әр жаңа бағанның не үшін қажет екенін бір сөйлеммен
түсіндіріңіз.

```python
# 1. Есептелген сандық баған: Қалалық және тасжолдағы жанармай шығынының айырмасы
df['MPG_Difference'] = df['highway MPG'] - df['city mpg']

# 2. Санаттық баған: Көлік бағасын (MSRP) pd.cut() арқылы санаттарға бөлу
bins = [0, 25000, 50000, 100000, float('inf')]
labels = ['Budget', 'Average', 'Premium', 'Luxury']
df['Price_Category'] = pd.cut(df['MSRP'], bins=bins, labels=labels)

# Нәтижені тексеру
print(df[['Make', 'Model', 'highway MPG', 'city mpg', 'MPG_Difference', 'MSRP', 'Price_Category']].tail())
```


# Жаңа бағандардың маңызы:

`'MPG_Difference'` (Сандық баған): Бұл баған көліктің тасжол мен қала ішіндегі жанармай тиімділігінің айырмасын көрсетеді, бұл қалалық көліктерді ұзақ жолға арналған модельдерден ажыратуға мүмкіндік береді.

`'Price_Category'` (Санаттық баған): Көліктерді бағалық сегменттерге топтастыру арқылы әртүрлі баға класындағы қозғалтқыш қуаты мен жанармай шығынының өзгеру заңдылықтарын талдауды жеңілдетеді.

# Тапсырма 7. Фильтрлеу
EDA бағытыңызға сай кемінде үш сұрақ қойып, оларға фильтрлеу арқылы жауап беріңіз. Бір
сұрақ бір шартпен, бір сұрақ `&` немесе `|` арқылы бірнеше шартпен, бір сұрақ `isin()` ,
`between()` немесе `query()` арқылы шешілуі керек. Әр сұрақты Markdown ұяшығында жазып,
нәтижесін санмен түсіндіріңіз.

```python
# 1-сұрақ: 'Luxury' баға санатына жататын қанша көлік бар және олардың негізгі деректері қандай?
luxury = df[df['Price_Category'] == 'Luxury']
print("Саны:", len(luxury))
print(luxury[['Make', 'Model', 'MSRP', 'Price_Category']].head())
```

```python
# 2-сұрақ (& немесе | арқылы бірнеше шартпен фильтрлеу): Тасжол мен қаладағы жанармай үнемділігінің
# айырмасы (MPG_Difference) 10-нан үлкен және баға санаты 'Budget' болатын қанша көлік бар?
eco_budget = df[(df['MPG_Difference'] > 10) & (df['Price_Category'] == 'Budget')]
print("Саны:", len(eco_budget))
print(eco_budget[['Make', 'Model', 'MPG_Difference', 'MSRP', 'Price_Category']].head())
```

```python
#3-сұрақ (isin(), between() немесе query() арқылы фильтрлеу): 'Average' немесе 'Premium' баға санатына кіретін
# және MSRP бағасы 30 000 мен 70 000 аралығында (between()) орналасқан және 'Engine HP' > 300 болатын қанша көлік бар?
mid_high = df[
    df['Price_Category'].isin(['Average', 'Premium']) &
    df['MSRP'].between(30000, 70000) &
    (df['Engine HP'] > 300)
]
print("Саны:", len(mid_high))
print(mid_high[['Make', 'Model', 'MSRP', 'Price_Category', 'Engine HP']].head())
```

# Тапсырма 8. Топтау
`groupby()` арқылы EDA бағытыңыздағы негізгі топтарды (мысалы, түр, санат, қала, класс)
кемінде екі сандық көрсеткіш бойынша салыстырыңыз. `agg()` арқылы бір кестеде кемінде үш
жиынтық функцияны ( `count` , `mean` , `median` , `min` , `max` , `std` ) қолданыңыз және нәтижені
сұрыптаңыз. Қай топтың ерекшеленетінін жазыңыз.

```python
df.groupby(["MSRP", "Model"])["Make"].count()
```

```python
summary_make = (
    df.groupby("Make", observed=False)
      .agg(
          model_count=("Model", "count"),
          valid_price_cat_count=("Price_Category", "count"),
          avg_msrp=("MSRP", "mean"),
          avg_hp=("Engine HP", "mean"),
      )
      .round(2)
      .sort_values(by="avg_msrp", ascending=False)
)

print(summary_make.head(10))
```

# Тапсырма 9. Екі белгі бойынша жиынтық кесте
`pd.pivot_table()` немесе екі баған бойынша `groupby()` арқылы екі белгінің бір көрсеткішке
бірге әсерін көрсететін кесте құрыңыз (мысалы, жыныс пен класс бойынша тірі қалу үлесі, аудан
мен бөлме түрі бойынша баға). Кестедегі ең жоғары және ең төмен мәндерді тауып, олардың
мағынасын түсіндіріңіз.

```python
# pd.pivot_table() арқылы жиынтық кесте құру
pivot_msrp = pd.pivot_table(
    df,
    values='MSRP',
    index='Price_Category',
    columns='Vehicle Size',
    aggfunc='mean',
    observed=False
).round(2)

print(pivot_msrp)
```

```python
# Ең жоғары мән және оның орны (санат пен өлшем)
max_val = pivot_msrp.max().max()
max_pos = pivot_msrp.stack().idxmax()

# Ең төмен мән (бос емес мәндер ішінен) және оның орны
min_val = pivot_msrp.min().min()
min_pos = pivot_msrp.stack().idxmin()

print(f"Ең жоғары орташа MSRP: {max_val} -> {max_pos}")
print(f"Ең төмен орташа MSRP: {min_val} -> {min_pos}")
```

Неге сонша жоғары: ірі өлшем (Large) + жоғары класс (Luxury) үйлесімі кезінде қозғалтқыш қуаты, интерьер сапасы және қосымша опциялар бағаны күрт өсіреді, әрі топ ішіндегі өте қымбат үлгілер орташа мәнді (mean) одан әрі жоғары тартады.

Неге ірі өлшем арзан: деректер жиынтығында бұрынғы жылдардағы (мысалы, 1990-жылдар немесе 2000-жылдардың басы) ескі үлкен пикаптар, фургондар немесе тозығы жеткен/бастапқы бағасы төмен үлкен көліктер кездесуі мүмкін. Олардың көлемі үлкен (Large) болғанымен, номиналды MSRP бағасы өте төмен тіркелген.

# Тапсырма 10. Нәтижелерді сақтау және қорытынды
Тазаланған кестені және ең маңызды жиынтық кестені `to_csv()` арқылы CSV файлдарына
сақтаңыз. Ноутбуктың соңында 5–7 сөйлемнен тұратын қорытынды жазыңыз: EDA бағытыңыз
бойынша үш негізгі тұжырым (сандармен), деректердің шектеулері (бос мәндер, кезең,
іріктеме) және келешекте тағы нені тексеруге болатыны.

```python
# 1. Тазаланған / жаңа бағандары бар негізгі кестені сақтау
# index=False — артық индекс бағанын файлға жазбау үшін
df.to_csv("cleaned_car_data.csv", index=False, encoding="utf-8-sig")

# 2. Ең маңызды жиынтық кестені (мысалы, Price_Category бойынша summary) сақтау
# Мұнда индекс (Price_Category) маңызды болғандықтан index=True қалдырылады
summary = df.groupby("Price_Category", observed=False).agg(
        count=("MSRP", "count"),
        mean_msrp=("MSRP", "mean"),
        median_msrp=("MSRP", "median"),
        mean_mpg_diff=("MPG_Difference", "mean"),
    ).round(2)

summary.to_csv("summary_by_price_category.csv", index=True, encoding="utf-8-sig")

print("Файлдар сәтті сақталды: cleaned_car_data.csv және summary_by_price_category.csv")
```
