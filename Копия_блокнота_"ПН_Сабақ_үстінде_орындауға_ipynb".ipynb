### Тапсырма 1. Тізім элементтері

`cities = ["Almaty", "Astana", "Shymkent", "Karaganda", "Aktobe"]` тізімін құрыңыз. Тізімнің бірінші элементін, соңғы элементін (теріс индекс арқылы) және ұзындығын шығарыңыз.

### Тапсырма 2. Сандар тізімі

`numbers = [4, 8, 15, 16, 23, 42]` тізімі берілген. `sum()`, `max()`, `min()` және `len()` функциялары арқылы сандардың қосындысын, ең үлкенін, ең кішісін және орташа мәнін шығарыңыз.

**Күтілетін нәтиже:**

```
108
42
4
18.0
```

````python
# Шешіміңізді осында жазыңыз
numbers = [4, 8, 15, 16, 23, 42]
print(sum(numbers))
print(max(numbers))
print(min(numbers))
print(sum(numbers)/len(numbers))
````

### Тапсырма 3. Элемент қосу және өшіру

`fruits = ["apple", "banana", "cherry"]` тізімінің соңына `"kiwi"` элементін `append()` арқылы, басына `"mango"` элементін `insert()` арқылы қосыңыз. Содан кейін `"banana"` элементін `remove()` арқылы өшіріңіз. Әр қадамнан кейін тізімді шығарыңыз.

**Күтілетін нәтиже:**

```
['apple', 'banana', 'cherry', 'kiwi']
['mango', 'apple', 'banana', 'cherry', 'kiwi']
['mango', 'apple', 'cherry', 'kiwi']
```

````python
# Шешіміңізді осында жазыңыз
fruits = ["apple", "banana", "cherry"]
fruits.append('kiwi')
print(fruits)
fruits.insert(0, 'mango')
print(fruits)
fruits.remove('banana')
print(fruits)
````

### Тапсырма 4. Сұрыптау

`scores = [78, 92, 65, 88, 54, 99, 71]` тізімін өсу ретімен және кему ретімен сұрыптап шығарыңыз. Кесінді алу арқылы ең жоғары үш баллды шығарыңыз.

**Күтілетін нәтиже:**

```

[99, 92, 88, 78, 71, 65, 54]
[99, 92, 88]
```

````python
# Шешіміңізді осында жазыңыз
scores = [78, 92, 65, 88, 54, 99, 71]
scores.sort()
print(scores)
scores.sort(reverse= True)
print(scores)
print(scores[:3])
````

### Тапсырма 5. Тізім бойынша цикл

`names = ["Aigerim", "Nurlan", "Dana", "Arman"]` тізімінің элементтерін `enumerate()` арқылы `1. Aigerim` форматында нөмірлеп шығарыңыз.

**Күтілетін нәтиже:**

```
1. Aigerim
2. Nurlan
3. Dana
4. Arman
```

````python
# Шешіміңізді осында жазыңыз
names = ["Aigerim", "Nurlan", "Dana", "Arman"]
for i, name in enumerate(names, start= 1):
  print(f'{i}. {name}')
````

### Тапсырма 6. Тізімді толтыру

## Анықтама: мәліметтер құрылымдарының әдістері

Төмендегі кестелерде тізімдерге, кортеждерге, жиындарға және сөздіктерге қолданылатын негізгі әдістер мен функциялар жинақталған. Мысалдардағы `a`, `t`, `s` және `d` сәйкесінше тізімді, кортежді, жиынды және сөздікті білдіреді.

### Тізім (list) әдістері

Тізім өзгертілетін құрылым: оның әдістерінің көбі тізімнің өзін өзгертеді және ештеңе қайтармайды (`None`).

| Әдіс | Сипаттамасы | Мысал | Нәтиже |
|---|---|---|---|
| `append(x)` | соңына бір элемент қосады | `a = [1, 2]; a.append(3)` | `[1, 2, 3]` |
| `insert(i, x)` | `i` индексіне элемент қосады | `a = [1, 3]; a.insert(1, 2)` | `[1, 2, 3]` |
| `extend(other)` | басқа тізімнің элементтерін соңына қосады | `a = [1]; a.extend([2, 3])` | `[1, 2, 3]` |
| `remove(x)` | бірінші кездескен `x` мәнін өшіреді | `a = [1, 2, 1]; a.remove(1)` | `[2, 1]` |
| `pop()` | соңғы элементті алып шығады және қайтарады | `a = [1, 2, 3]; a.pop()` | `3`, тізім `[1, 2]` |
| `pop(i)` | `i` индексіндегі элементті алып шығады | `a = [1, 2, 3]; a.pop(0)` | `1`, тізім `[2, 3]` |
| `clear()` | тізімді толық тазалайды | `a = [1, 2]; a.clear()` | `[]` |
| `index(x)` | `x` мәнінің бірінші индексін қайтарады | `[5, 7, 5].index(7)` | `1` |
| `count(x)` | `x` мәнінің неше рет кездесетінін қайтарады | `[5, 7, 5].count(5)` | `2` |
| `sort()` | тізімнің өзін өсу ретімен сұрыптайды | `a = [3, 1, 2]; a.sort()` | `[1, 2, 3]` |
| `sort(reverse=True)` | тізімнің өзін кему ретімен сұрыптайды | `a = [3, 1, 2]; a.sort(reverse=True)` | `[3, 2, 1]` |
| `reverse()` | элементтердің ретін керісінше айналдырады | `a = [1, 2, 3]; a.reverse()` | `[3, 2, 1]` |
| `copy()` | тізімнің тәуелсіз көшірмесін қайтарады | `b = a.copy()` | жаңа тізім |

### Кортеж (tuple) әдістері

Кортеж өзгермейтін болғандықтан, оның тек екі әдісі бар. Элемент қосатын, өшіретін немесе ауыстыратын әдістер кортежде жоқ.

| Әдіс | Сипаттамасы | Мысал | Нәтиже |
|---|---|---|---|
| `index(x)` | `x` мәнінің бірінші индексін қайтарады | `(5, 7, 5).index(7)` | `1` |
| `count(x)` | `x` мәнінің неше рет кездесетінін қайтарады | `(5, 7, 5).count(5)` | `2` |

| Амал | Сипаттамасы | Мысал | Нәтиже |
|---|---|---|---|
| бөліп алу | элементтерді айнымалыларға меншіктеу | `x, y = (3, 7)` | `x = 3`, `y = 7` |
| бір элементті кортеж | үтір міндетті | `t = (5,)` | `(5,)` |
| түрлендіру | тізімнен кортеж және керісінше | `tuple([1, 2])`, `list((1, 2))` | `(1, 2)`, `[1, 2]` |

### Жиын (set) әдістері

Жиын тек бірегей элементтерді сақтайды, элементтердің реті жоқ, сондықтан индекс арқылы жүгінуге болмайды. Бос жиын `set()` арқылы құрылады, ал `{}` бос сөздікті білдіреді.

| Әдіс | Сипаттамасы | Мысал | Нәтиже |
|---|---|---|---|
| `add(x)` | бір элемент қосады | `s = {1, 2}; s.add(3)` | `{1, 2, 3}` |
| `update(other)` | бірнеше элемент қосады | `s = {1}; s.update([2, 3])` | `{1, 2, 3}` |
| `remove(x)` | элементті өшіреді, жоқ болса `KeyError` береді | `s = {1, 2}; s.remove(2)` | `{1}` |
| `discard(x)` | элементті өшіреді, жоқ болса қате бермейді | `s = {1, 2}; s.discard(5)` | `{1, 2}` |
| `pop()` | кездейсоқ элементті алып шығады | `s.pop()` | бір элемент |
| `clear()` | жиынды толық тазалайды | `s.clear()` | `set()` |
| `union(other)` немесе `\|` | бірігу: екі жиынның барлық элементі | `{1, 2} \| {2, 3}` | `{1, 2, 3}` |
| `intersection(other)` немесе `&` | қиылысу: ортақ элементтер | `{1, 2} & {2, 3}` | `{2}` |
| `difference(other)` немесе `-` | айырма: тек бірінші жиындағы элементтер | `{1, 2} - {2, 3}` | `{1}` |
| `symmetric_difference(other)` немесе `^` | тек бір жиында ғана бар элементтер | `{1, 2} ^ {2, 3}` | `{1, 3}` |
| `issubset(other)` | барлық элемент екінші жиында бар ма | `{1}.issubset({1, 2})` | `True` |
| `issuperset(other)` | екінші жиынның барлық элементі осында бар ма | `{1, 2}.issuperset({1})` | `True` |
| `isdisjoint(other)` | ортақ элемент жоқ па | `{1}.isdisjoint({2})` | `True` |

### Сөздік (dict) әдістері

Сөздік "кілт: мән" жұптарын сақтайды. Кілттер бірегей және өзгермейтін типте (жол, сан, кортеж) болуы тиіс. Мәнге `d[key]` арқылы жүгінгенде кілт жоқ болса, `KeyError` қатесі шығады.

| Әдіс | Сипаттамасы | Мысал | Нәтиже |
|---|---|---|---|
| `d[key] = value` | жұп қосады немесе мәнді өзгертеді | `d = {}; d["a"] = 1` | `{"a": 1}` |
| `get(key)` | мәнді қайтарады, кілт жоқ болса `None` | `{"a": 1}.get("b")` | `None` |
| `get(key, default)` | кілт жоқ болса әдепкі мәнді қайтарады | `{"a": 1}.get("b", 0)` | `0` |
| `keys()` | барлық кілтті қайтарады | `{"a": 1, "b": 2}.keys()` | `dict_keys(['a', 'b'])` |
| `values()` | барлық мәнді қайтарады | `{"a": 1, "b": 2}.values()` | `dict_values([1, 2])` |
| `items()` | барлық "кілт, мән" жұбын қайтарады | `{"a": 1}.items()` | `dict_items([('a', 1)])` |
| `update(other)` | бірнеше жұпты қосады немесе жаңартады | `d = {"a": 1}; d.update({"b": 2})` | `{"a": 1, "b": 2}` |
| `pop(key)` | жұпты өшіреді және мәнін қайтарады | `d = {"a": 1}; d.pop("a")` | `1`, сөздік `{}` |
| `pop(key, default)` | кілт жоқ болса қате бермей, әдепкі мәнді қайтарады | `{}.pop("a", None)` | `None` |
| `setdefault(key, default)` | кілт жоқ болса қосады, мәнін қайтарады | `d = {}; d.setdefault("a", [])` | `[]`, сөздік `{"a": []}` |
| `del d[key]` | жұпты өшіреді | `d = {"a": 1}; del d["a"]` | `{}` |
| `clear()` | сөздікті толық тазалайды | `d.clear()` | `{}` |
| `copy()` | сөздіктің көшірмесін қайтарады | `d2 = d.copy()` | жаңа сөздік |

### Ортақ функциялар мен операторлар

Төмендегі функциялар мен операторлар бірнеше құрылымға бірдей қолданылады. Белгілері: Т — тізім, К — кортеж, Ж — жиын, С — сөздік.

| Функция немесе амал | Сипаттамасы | Қолданылады | Мысал | Нәтиже |
|---|---|---|---|---|
| `len(x)` | элементтер саны | Т, К, Ж, С | `len([1, 2, 3])` | `3` |
| `x in y` | элементтің бар-жоғын тексеру (сөздікте кілт тексеріледі) | Т, К, Ж, С | `2 in {1, 2}` | `True` |
| `sum(x)` | сандардың қосындысы | Т, К, Ж | `sum([1, 2, 3])` | `6` |
| `max(x)`, `min(x)` | ең үлкен және ең кіші элемент | Т, К, Ж, С | `max((4, 9, 1))` | `9` |
| `sorted(x)` | сұрыпталған жаңа тізім қайтарады, бастапқыны өзгертпейді | Т, К, Ж, С | `sorted({3, 1, 2})` | `[1, 2, 3]` |
| `x[i]` | индекс бойынша элемент | Т, К | `[10, 20, 30][1]` | `20` |
| `x[start:stop]` | кесінді алу | Т, К | `[1, 2, 3, 4][1:3]` | `[2, 3]` |
| `+`, `*` | біріктіру және қайталау | Т, К | `[1] + [2]`, `(0,) * 3` | `[1, 2]`, `(0, 0, 0)` |
| `enumerate(x, start=1)` | элементті реттік нөмірімен бірге береді | Т, К, Ж, С | `list(enumerate(["a", "b"], 1))` | `[(1, 'a'), (2, 'b')]` |
| `zip(x, y)` | екі тізбекті жұптайды | Т, К | `list(zip([1, 2], ["a", "b"]))` | `[(1, 'a'), (2, 'b')]` |
| `list()`, `tuple()`, `set()`, `dict()` | бір құрылымды басқаға түрлендіру | барлығы | `set([1, 1, 2])` | `{1, 2}` |
